prpgrams start in the main() function, the function keyword is fn

```rust
fn main() {
    println!("Hello World!");
}
main();
```

print!() and println!() is pretty much the same, but println!() adds a space at the end of the line

```rust
// All variables have to be initialised

fn main() {
    let x: i32;
    let y: i32; // give a warning for unused y
    let _z: i32; // prevents the warning by using _*

    x = 5;
    
    assert_eq!(x, 5);
    println!("Success!");
}

main();
```

```rust
// variables are immutable by default, you have to let them be mutated

fn main() {
    let mut x: i32 = 1;
    x = x + 2;
    x+=3;

    assert_eq!(x, 6);
    println!("Success!")
}

main();
```

```rust
// scopes

fn main() {
    let x: i32 = 10;
    let y: i32 = 6;

    {
        let y: i32 = 5; 
        println!("The value of x is {}, and the value of y is {}", x, y);
    }
    println!("The value of x is {}, and the value of y now is {}", x, y);
    // y is out of scope here so the program throws an error if we do not declare
    // it at the start
}

main();
```

```rust
// use of functions

fn main() {
    define_x();
    
}

fn define_x() {
    let x: &str = "Hello";
    println!("{}, world", x);
}

main();
```

```rust
// Shadowing, have a variable with the same name

fn main() {
    let x: i32 = 5;

    {
        let x = 12;
        assert_eq!(x, 12);
    }

    assert_eq!(x, 5);

    let x = 42;

    println!("{}", x);
}

main();
```
