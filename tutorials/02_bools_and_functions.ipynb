```rust
// recap, char is 4 bytes, bool is 1 byte (not always), unit is an empty tuple of size 0

// a statement performs and action but does not produce a value (fn definitions, code that ends with ; usually)
// an expression evaluated to a value

fn main() {
    let x: u32 = 5u32;

    let y: u32 = {
        let x_squared = x * x;
        let x_cube = x_squared * x;

        x_cube + x_squared + x // this expression will be assigned to 'y'. NO ; for return
         // this is an expression, it produces a result
    }; // this is a statemenr

    let z: () = {
        2 * x; // here z is assigned a unit type, this block does not return a value
    };

    print_value(x);
    print_value(y);
    // print_value(z); this is not handled for now, later can create enum variable with both types or use Debug
}

fn print_value(a: u32) {
    println!("The value is {}", a);
}

main();
```

```rust
fn main() {
    let v: i32 = {
        let mut x: i32 = 1;
        x += 2; // cannot return this directly
        x
    };

    assert_eq!(v, 3);

    println!("Success");
}

main();
```

```rust
fn main() {
    never_return();

    println!("Failed!");
}

fn never_return() -> ! {
    panic!() // maked the program exit and return an error, a statement that should not be accessed

    // this is a diverging function
}

main();
```
