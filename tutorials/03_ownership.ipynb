```rust
// OWNERSHIP is a unique rust feature

// Set of rules that govern memory managment and evaluated at compile time.
// each value has an owner, there can only be one owner at a time, when the owner goes out of scope, it is dropped
// two types of regions in RAM used at runtime: stack memory and heap memory

// stack: last in, first out, data has a known fixed size, pushing to the stack is faster than allocating on the heap,
// location for new data is always at the top of the stack

// types of unknown size are allocated to the heap, and a pointer to the value is pushed to the stack
```

```rust
fn main() {
    let x = 42;
    let y = 10;
    let z = add_numbers(x, y);

    println!("The result is {}", z);
}

fn add_numbers(a: i32, b: i32) -> i32 {
    let c = a + b;
    c
}

main();

// main and local variables are pushed to the stack, so is add_numbers alongside all other variables
```

```rust
// heap: data of not known fixxed size, allocating on the heap returns a pointer. allocating and accesing data on the heap is slower
// For instance the String type (mutable, size can change at runtime, valuye on heap, pointer on stack)

fn main() {
    let s1 = String::from("hello");
    println!("THe value of the string is {}", s1)
}

main(); // s1 holds a pointer (but hello is printed and not the memory addess... hmm...)

// sq holds a pointer, the lenght, and capacity (total amount of memery received from the allocator)
// each has size u size (8 bytes / 64 bits), so s1 will have size 24 bytes, so size is known at compile time.

// in heap memory, the data is allocated in a contiguous block (we only need the memory address to the first element + len)
```

```rust
// Copy vs move

fn main() {
    let x = 5;
    let y = x; // gets copied to the stack
    // y and x point to different memory addesses, this is for a vairable that lives
    // in the stack memory, copying in the stack is cheap

    let s1 = String::from("hello");
    let s2 = s1; // just the pointer will get copied into s2, not the whole data on the heap!

    // both point to the same location on heap memory, this would violate the second rule (there can only be one owner!)
    // as a result, rust gets rid of s1

    // println!("s1: {} and s2: {},", s1, s2) // this gives an error

    println!("x: {} and y: {},", x, y);

    // you can do a deep copy on the heap (might be expensive)
    let s3 = String::from("hello");
    let s4 = s3.clone();

    println!("s3: {} and s4: {},", s3, s4);
}

main();
```

```rust
// Example of ownership with functions

fn main() {
    let s = String::from("hello"); // s comes into scope

    takes_ownership(s); // s value moves into the function, no longer valid here

    let x = 5; // x comes into scope

    makes_copy(x); // x moves into the function, but it is fixed size i32,
                    // so x here is still valid
    // let _f = s; // this line would give an error

    let _y = x; // this is fine since it is a copy
}

fn takes_ownership(some_string: String) {
    println!("{}", some_string);
}

fn makes_copy(some_integer: i32) {
    println!("{}", some_integer);
}

main();
```

```rust
fn main() {

    let s1 = gives_ownership();
    let s2 = String::from("hello");
    let s3 = takes_and_gives_back(s2); // s2 is dead :(

    println!("{}, {}", s1, s3);
     
}

fn gives_ownership() -> String {
    let some_string = String::from("hi");
    some_string
}

fn takes_and_gives_back(a_string: String) -> String {
    a_string
}

main();
```
