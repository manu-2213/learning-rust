print!() and println!() is pretty much the same, but println!() adds a space at the end of the line

```rust
// All variables have to be initialised

fn main() {
    let x: i32;
    let y: i32; // give a warning for unused y
    let _z: i32; // prevents the warning of not being used by using _*

    x = 5;
    
    assert_eq!(x, 5);
    println!("Success!");
}

main();
```

```rust
// variables are immutable by default, you have to let them be mutated

fn main() {
    let mut x: i32 = 1;
    x = x + 2;
    x+=3;

    assert_eq!(x, 6);
    println!("Success!")
}

main();
```

```rust
// scopes

fn main() {
    let x: i32 = 10;
    let y: i32 = 6;

    {
        let y: i32 = 5; 
        println!("The value of x is {}, and the value of y is {}", x, y);
    }
    println!("The value of x is {}, and the value of y now is {}", x, y);
    // y is out of scope here so the program throws an error if we do not declare
    // it at the start
}

main();
```

```rust
// use of functions

fn main() {
    define_x();
    
}

fn define_x() {
    let x: &str = "Hello";
    println!("{}, world", x);
}

main();
```

```rust
// Shadowing, have a variable with the same name

fn main() {
    let x: i32 = 5;

    {
        let x = 12;
        assert_eq!(x, 12);
    }

    assert_eq!(x, 5);

    let x = 42;

    println!("{}", x);
}

main();
```

```rust
#[allow(unused_variables)] // Tell the compiler this is allowed

fn main() {
    let mut x: i32 = 123;
    let mut z: i64 = 1_000_000;
    println!("The variable is {}", x);
}

main();
```

```rust
fn main() {
    let (mut x, y) = (1, 2); // mut has to be per element in tuple
    x += 2;
    assert_eq!(x, 3);
    assert_eq!(y, 2);
    println!("Success!");
}

main();
```

```rust
fn main() {
    let (x, y); // declare two variables at once
    (x,..) = (3, 4); // another way of assigning variables
    [..,y] = [1, 2];

    assert_eq!([x, y], [3, 2]);
    println!("Success!");
}

main();
```

```rust
// handling ints, passing a float will make the compiler complain here
// default times are i32 and f64 for ints and floats

fn main() {
    let x: isize = -123; // signed integer with size depending on computer architecture
    let y: usize = 124; // same but unsigned

    println!("The numbers are {}, and {}", x, y);
}

main();
```

```rust
// floating points f32 and f64
// representation according to IEEE-754
// if we do not assign a type to a variable, the compiler infers it

fn main() {
    let x: i32 = 5;
    let mut y: u32 = 5;

    y = x; // This would fail

    let a: i32 = 5;
    let mut b = 5;

    b = a; // This works since the default integer type is i32
    
}
```

```rust
fn main() {
    //assert!(0.1 + 0.2 == 0.3); // problem with floating point precision
    assert!(0.1 as f32 + 0.2 as f32 == 0.3 as f32);
    assert!(0.1_f32 + 0.2_f32 == 0.3_f32);

    println!("Success!");
}

main();
```

```rust
fn main() {
    let mut sum: i32 = 0;
    for i in -3..2 {
        sum += i // like python range(-3, 2), the last value is not used
    }

    assert!(sum == -5);

    for c in 'a'..='f' { // this way we include the last element
        println!("{}", c);
    }

    for c in 'a'..='f' { // this way we include the last element
        println!("{}", c as u8); // get the ASCII binary representation
    }
}

main();
```

```rust
use std::ops::{Range, RangeInclusive};

fn main() {
    assert_eq!((1..5), Range{start:1, end:5});
    assert_eq!((1..=5), RangeInclusive::new(1, 5)); //funnily enough (1..6) throws an error here

    println!("Success!");
}

main();

// (1..5) is of type Range<i32>
// (1..=5) is of type RangeInclusive<i32>
// Range only implements == against another Range type
```
